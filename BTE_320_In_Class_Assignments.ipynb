```python
#Income Statement
Sales=float(input("Enter Company Sales"))
COGS=Sales*.4
GrossProfit=Sales-COGS

#Enhance
TaxR=float(input("Enter Tax Rate in %:"))
Taxes=TaxR*GrossProfit
NetIncome=GrossProfit-Taxes

print()

print("Company XYZ Income Statement:")
print(f"Sales:                 $ {Sales}")
print(f"Cost of Goods Sold:    $ {COGS}")
print(f"Gross Profit:          $ {GrossProfit}")
print(f"Taxes:                 $ {Taxes}")
print(f"Net Income:            $ {NetIncome}")

```
